<h3>1. Un programa que convierta grados Celsius a Farenheint.</h3>

```python
Celsius = float(input("Ingresa la temperatura en grados Celsius: "))
Fahrenheit = (Celsius * 9/5) + 32
print(f"La temperatura en Fahrenheit es: {Fahrenheit}°F")
```

<h3>2. Un programa que convierta Kilómetros a millas.</h3>

```python
Kilometros = float(input("Ingresa la distancia en kilómetros: "))
Millas = Kilometros * 0.621371
print(f"La distancia en millas es: {Millas}")
```

<h3> 3. Un programa que calcule la propina de una cena. </h3>

```python
Cuenta=float(input("Ingresa el total de la cuenta es de $"))
Porcentaje=float(input("Ingresa el porcentaje que deseas dar de propina (Ej. 0.10, 0.20, etc) es de "))
Propina=Cuenta*Porcentaje
Total_a_pagar = Cuenta + Propina
print(f"Su total a pagar es de ${Total_a_pagar}")
```
