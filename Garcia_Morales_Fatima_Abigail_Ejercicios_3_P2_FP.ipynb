# **PROBLEMA: 2**
Un programa que determine el Máximo Común Denominador (MCD) de dos números.

### **Variables:**

1. a (Int): Primer número ingresado.

2. b (Int): Segundo número ingresado.

3. temp (Int): Variable auxiliar utilizada para realizar el intercambio de valores durante el algoritmo de Euclides.

### **Descripcion en lenguaje natural.**

1. Definir las variables a, b, temp.

2. Leer los dos números enteros positivos y asignarlos a a y b.

3. Mientras b sea diferente de cero, calcular el residuo de a entre b, actualizar los valores mediante el Algoritmo de Euclides (temp = b, b = a % b, a = temp) y repetir el ciclo.

4. Mostrar en pantalla el valor almacenado en a como el Máximo Común Denominador (MCD).

### **Diagrama de flujo.**
Captura de pantalla 2026-11-03 202113.png

### **Código y resultado.**

```python
a = int(input("Ingrese el primer número entero: "))
b = int(input("Ingrese el segundo número entero: "))
while b != 0:
    temp = b
    b = a % b
    a = temp
print(f"El Máximo Común Denominador (MCD) es: {a}")
```
