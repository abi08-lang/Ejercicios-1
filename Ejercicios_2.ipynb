**<h3>PROBLEMA 1 </h3>** <h3>Programa para calcular el precio final de un producto que tiene tiene un descuento. (Solicitar el % del Descuento)</h3>

```python
precio = float(input("Ingrese el precio del producto: "))
descuento_porcentaje = float(input("Ingrese el porcentaje de descuento (Ej. 10, 12, 15, etc.): "))
monto_descuento = precio * (descuento_porcentaje / 100)
precio_final = precio - monto_descuento
print(f"Descuento aplicado: {monto_descuento:.2f}")
print(f"Precio final a pagar: {precio_final:.2f}")
```

**<h3>PROBLEMA 2 </h3>** <h3>Programa para convertir Radianes a grados. </h3>

```python
import math
radianes = float(input("Ingrese valor en radianes: "))
grados = radianes * (180 / math.pi)
print(f"Valor en grados: {grados:.2f}°")
```

**import math** se utiliza para poder acceder a las funciones y constantes matemáticas de la librería math. Dentro de ella encontramos **math.pi**, que representa el número π y se usa en cálculos de geometría y trigonometría. Cuando queremos mostrar ese valor en pantalla con solo dos decimales, aplicamos el formato **.2f**, que redondea y limita la salida a dos cifras después del punto decimal.
