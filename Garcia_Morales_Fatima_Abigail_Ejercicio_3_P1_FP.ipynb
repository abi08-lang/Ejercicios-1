## **PROBLEMA 1**
### Programa que lea 3 números  y determine cual es el mayor.

## **Variables:**

**num1 (float):** Almacena el primer número ingresado por el usuario.

**num2 (float):** Almacena el segundo número ingresado por el usuario.

**num3 (float):** Almacena el tercer número ingresado por el usuario.

**mayor:** Almacena temporalmente el valor del número mayor.

## **Descripción en lenguaje natural.**

1. Definir las variables num1, num2, num3 y mayor.

2. Leer los tres números y asignarlos a num1, num2 y num3.

3. Asignar temporalmente el valor de num1 a la variable mayor.

4. Actualizar el valor de mayor si num2 o num3 son mayores que él.

5. Mostrar en pantalla el valor almacenado en mayor.

## **Diagrama de flujo.**
Captura de pantalla 2026-11-03 193413.png

## **Código y resultado.**

```python
num1 = float(input("Ingrese el primer número: "))
num2 = float(input("Ingrese el segundo número: "))
num3 = float(input("Ingrese el tercer número: "))
mayor = num1
if num2 > mayor:
    mayor = num2
if num3 > mayor:
    mayor = num3
print(f"El número mayor es: {mayor}")
```
